```python
from transformers import GPT2Tokenizer, GPT2Model
import struct
from pathlib import Path
```

```python
model = GPT2Model.from_pretrained('gpt2')
bin_file_name = "model_weights.bin"
```

```python
# Deletes the file; ignores the error if the file doesn't exist
Path(bin_file_name).unlink(missing_ok=True)

weights_dict = model.state_dict()
```

```python
byte_offset_index = dict()
```

```python
offset = 0
for layer_name, weights in weights_dict.items():
    print(layer_name, weights.shape)
    val = weights.flatten()
    val_np = val.tolist()    
    byte_offset_index[layer_name] = offset
    offset = offset + len(val_np)
    bin_data = struct.pack(f"{len(val_np)}f", *val_np)
    with open(bin_file_name, "ab") as file:
        file.write(bin_data)
```
